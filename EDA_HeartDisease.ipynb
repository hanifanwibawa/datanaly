```python
from ucimlrepo import fetch_ucirepo

# fetch dataset
heart_disease = fetch_ucirepo(id=45)

# data (as pandas dataframes)
X = heart_disease.data.features #supervised
y = heart_disease.data.targets #supervised

# metadata
print(heart_disease.metadata)

# variable information
print(heart_disease.variables)
```

```python
import pandas as pd

df = pd.concat([X, y], axis=1)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 15)) # ukuran
for i, col in enumerate(df.columns, 1): # looping satu per satu
  plt.subplot(5, 4, i) # menampilkan 5 kebawah, 4 ke kanan, i adalah datanya
  sns.countplot(data=df, x=col) # menampilkan visual data, df adalah data yg disebutkan diatas, sumbu X adalah col
  plt.title(f'{col} distribution') # menampilkan judul (col + "distribution")
  plt.tight_layout() # bagaimana cara menampilkan datanya

plt.show()
```

```python
# menampilkan 1 data saja

plt.figure(figsize=(4, 4))

df['sex_label'] = df['sex'].map({0: 'Cewe', 1: 'Cowo'})
data = df['sex_label'].value_counts() # value_counts sex cowo berapa, cewe berapa

plt.pie(data, labels=data.index)
plt.title('target distribution')
plt.tight_layout()
```
